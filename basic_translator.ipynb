```python

#Language: vouwle => 'g'

def translate(phrase):
    translation = ""
    for letter in phrase:
        if  letter in "AEIOUaeuio":
            if letter.isupper():
                translation = translation + "G"
            else:
                translation = translation + "g"
        else:
            translation = translation + letter
    return translation

translate(input("Enter a phrase: "))

```
